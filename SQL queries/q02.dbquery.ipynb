```sql
create or replace view q02 as 
select h.hospital_id, 
h.hospital_name,
round(sum(case when f.status = 'settled' then f.approved_amount else 0 end),2) as total_paid,
count(case when f.status = 'settled' then 1 else 0 end) as settled_claims
from fact_claims f
join dim_hospital h
on f.hospital_sk = h.hospital_sk
group by h.hospital_id, h.hospital_name, f.status
order by total_paid desc
limit 10
```
