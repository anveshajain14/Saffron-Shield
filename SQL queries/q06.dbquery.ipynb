```sql
create or replace view q06 as 
select
p.plan_code,
count(*) as claim_count,
sum(case when f.room_rent_over_cap = true then 1 else 0 end) as over_cap_count,
round(over_cap_count/ count(*), 4) as over_cap_share
from fact_claims f
join dim_plan p
on f.plan_sk = p.plan_sk
group by p.plan_code
```
