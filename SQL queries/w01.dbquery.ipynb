```sql
create or replace view w01 as 
SELECT
  pl.plan_code,
  d.month_start,
  ROUND(SUM(case when f.status = 'settled' then f.approved_amount else 0 end),2)AS paid_amount,
  ROUND(SUM(pl.annual_premium) / 12.0, 2) AS earned_premium,
  ROUND(
    SUM(case when f.status = 'settled' then f.approved_amount else 0 end) FILTER (WHERE f.status = 'settled')
    / NULLIF(SUM(pl.annual_premium) / 12.0, 0),
    2
  ) AS loss_ratio
FROM claims.analytics.fact_claims f
JOIN claims.analytics.dim_plan pl ON f.plan_sk = pl.plan_sk
JOIN claims.analytics.dim_date d ON f.admission_date_key = d.date_key
GROUP BY pl.plan_code, d.month_start;
```
