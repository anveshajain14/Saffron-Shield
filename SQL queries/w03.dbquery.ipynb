```sql
CREATE OR REPLACE VIEW w03 AS
SELECT
  h.hospital_id,
  h.city_tier,
  ROUND(SUM(f.approved_amount) FILTER (WHERE f.status = 'settled'), 2) AS total_paid,
  RANK() OVER (
    PARTITION BY h.city_tier
    ORDER BY SUM(f.approved_amount) FILTER (WHERE f.status = 'settled') DESC
  ) AS rank_in_tier
FROM fact_claims f
JOIN dim_hospital h ON f.hospital_sk = h.hospital_sk
GROUP BY h.hospital_id, h.city_tier;
```
