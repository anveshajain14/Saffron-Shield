```sql
CREATE OR REPLACE VIEW claims.analytics.r8_provider_join AS
SELECT
  p.hospital_id,
  h.hospital_name,
  h.city_tier,
  p.accreditation,
  p.accredited_until,
  COALESCE(COUNT(CASE 
    WHEN p.accredited_until IS NOT NULL 
     AND c.admission_date_key > CAST(DATE_FORMAT(p.accredited_until, 'yyyyMMdd') AS INT) 
    THEN 1 
  END), 0) AS claims_after_lapse
FROM claims_lakebase.public.provider_directory p
LEFT JOIN claims.analytics.dim_hospital h
  ON h.hospital_id = p.hospital_id
LEFT JOIN claims.analytics.fact_claims c
  ON c.hospital_sk = h.hospital_sk
GROUP BY p.hospital_id, h.hospital_name, h.city_tier, p.accreditation, p.accredited_until
```
